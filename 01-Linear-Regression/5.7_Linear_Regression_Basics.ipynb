# 5.7 Linear Regression Basics

**Objective:** Simple Linear Regression using Salary Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import os

# Use THIS notebook's folder paths
NOTEBOOK_DIR = os.getcwd()
DATASETS_DIR = os.path.join(NOTEBOOK_DIR, 'datasets')
OUTPUTS_DIR = os.path.join(NOTEBOOK_DIR, 'outputs')
os.makedirs(OUTPUTS_DIR, exist_ok=True)

print(f"📁 Datasets: {DATASETS_DIR}")
print(f"📁 Outputs: {OUTPUTS_DIR}")
```

```python
# Load Dataset from datasets/ folder
df = pd.read_csv(os.path.join(DATASETS_DIR, 'Salary_Data.csv'))
X = df[['YearsExperience']].values
y = df['Salary'].values

print("Dataset Shape:", df.shape)
print(df.head(10))
```

```python
# Train Model
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

print("="*50)
print("RESULTS")
print("="*50)
print(f"Equation: Salary = {model.intercept_:.2f} + {model.coef_[0]:.2f} × YearsExperience")
print(f"R² Score: {r2_score(y, y_pred):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y, y_pred)):.2f}")
```

```python
# Save Output Image to outputs/ folder
plt.figure(figsize=(9, 5))
plt.scatter(df['YearsExperience'], df['Salary'], color='blue', alpha=0.7, s=70, label='Actual')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary ($)')
plt.title('5.7: Linear Regression - Salary Prediction')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUTS_DIR, '5.7_Linear_Regression_Basics.png'), dpi=120)
plt.show()
print(f"✅ Saved to: {OUTPUTS_DIR}/5.7_Linear_Regression_Basics.png")
```
